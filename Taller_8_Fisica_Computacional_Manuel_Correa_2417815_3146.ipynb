## 6. Taller 8: la radiación de cuerpo negro


### A. Enunciado y requisitos

La radiancia espectral de un cuerpo negro a temperatura $T$ está dada por la **ley de Planck**:

$$B_\lambda(\lambda, T) = \frac{2hc^2}{\lambda^5}\,\frac{1}{e^{hc/(\lambda k_B T)} - 1} ,$$

donde $B_\lambda$ es la potencia emitida por unidad de área, de ángulo sólido y de longitud de onda [W m⁻² sr⁻¹ m⁻¹], $\lambda$ la longitud de onda [m], $T$ la temperatura [K], $h = 6.62607015\times10^{-34}$ J s la constante de Planck, $c = 299792458$ m/s la rapidez de la luz en el vacío y $k_B = 1.380649\times10^{-23}$ J/K la constante de Boltzmann. El factor $2hc^2/\lambda^5$ cuenta los modos electromagnéticos con su energía y el factor con el exponencial es la ocupación de esos modos. Estas tres constantes son exactas en el Sistema Internacional.

**Variable adimensional.** Con $x = hc/(\lambda k_B T)$, la razón entre la energía de un fotón y la energía térmica, el espectro queda

$$B_\lambda \propto \frac{x^5}{e^x - 1} .$$

**Ley de desplazamiento de Wien.** El máximo del espectro cumple $dB_\lambda/dx = 0$, que se reduce a

$$x = n\left(1 - e^{-x}\right), \qquad n = 5 ,$$

ecuación trascendente sin solución en funciones elementales, con la raíz trivial $x = 0$ y una raíz positiva $x_5$. Esa raíz fija la constante de Wien $b$:

$$\lambda_{\max}\,T = b = \frac{hc}{k_B\,x_5} .$$

Si el espectro se expresa por unidad de frecuencia, $B_\nu \propto x^3/(e^x - 1)$, y el máximo cumple la misma ecuación con $n = 3$.

**Ley de Stefan-Boltzmann.** La potencia total emitida por unidad de área es $M = \int_0^\infty \pi B_\lambda\,d\lambda = \sigma T^4$, con

$$\sigma = \frac{2\pi k_B^4}{h^3 c^2}\,I, \qquad I = \int_0^\infty \frac{x^3}{e^x - 1}\,dx ,$$

donde $\sigma$ es la constante de Stefan-Boltzmann [W m⁻² K⁻⁴] e $I$ una integral pura, cuyo valor exacto es $\pi^4/15$. Es una integral impropia. El cambio de variable $x = t/(1 - t)$, con $dx = dt/(1-t)^2$, la lleva al intervalo finito $[0, 1]$:

$$I = \int_0^1 \frac{x(t)^3}{e^{x(t)} - 1}\,\frac{dt}{(1 - t)^2} .$$

##### MODULO

```python
%%writefile modulo_ceros.f90
module modulo_ceros
  implicit none
  integer, parameter :: dp = kind(1.0d0)
  integer, parameter :: max_iter_ceros = 100

contains

  real(dp) function F_wien(x, n_param)
    real(dp), intent(in) :: x, n_param
    if (x > 700.0_dp) then
       F_wien = x - n_param
    else
       F_wien = x - n_param * (1.0_dp - exp(-x))
    end if
  end function F_wien

  real(dp) function dF_wien(x, n_param)
    real(dp), intent(in) :: x, n_param
    if (x > 700.0_dp) then
       dF_wien = 1.0_dp
    else
       dF_wien = 1.0_dp - n_param * exp(-x)
    end if
  end function dF_wien

  subroutine biseccion(a_ini, b_ini, n_param, tol, raiz, n_iter, unidad_hist, id_metodo)
    real(dp), intent(in) :: a_ini, b_ini, n_param, tol
    integer, intent(in) :: unidad_hist, id_metodo
    real(dp), intent(out) :: raiz
    integer, intent(out) :: n_iter
    real(dp) :: a, b, c, fa, fc
    integer :: k

    a = a_ini
    b = b_ini
    fa = F_wien(a, n_param)
    if (fa * F_wien(b, n_param) >= 0.0_dp) then
       stop 'biseccion: no hay cambio de signo en el intervalo inicial'
    end if

    n_iter = 0
    raiz = 0.5_dp * (a + b)
    do k = 1, max_iter_ceros
       c = 0.5_dp * (a + b)
       n_iter = k
       raiz = c
       if (unidad_hist /= 0) then
          write(unidad_hist, '(2i6, es24.16)') id_metodo, k, c
       end if
       if ((b - c) <= tol) return
       fc = F_wien(c, n_param)
       if (fa * fc <= 0.0_dp) then
          b = c
       else
          a = c
          fa = fc
       end if
    end do
  end subroutine biseccion

  subroutine newton_raphson(x0, n_param, tol, raiz, n_iter, unidad_hist, id_metodo)
    real(dp), intent(in) :: x0, n_param, tol
    integer, intent(in) :: unidad_hist, id_metodo
    real(dp), intent(out) :: raiz
    integer, intent(out) :: n_iter
    real(dp) :: x, fx, dfx, paso
    integer :: k

    x = x0
    n_iter = 0
    raiz = x
    do k = 1, max_iter_ceros
       fx = F_wien(x, n_param)
       dfx = dF_wien(x, n_param)
       if (abs(dfx) <= tiny(1.0_dp)) then
          stop 'newton_raphson: derivada nula'
       end if
       paso = -fx / dfx
       x = x + paso
       n_iter = k
       raiz = x
       if (unidad_hist /= 0) then
          write(unidad_hist, '(2i6, es24.16)') id_metodo, k, x
       end if
       if (abs(paso) < tol) return
    end do
  end subroutine newton_raphson

  subroutine secante(x0_ini, x1_ini, n_param, tol, raiz, n_iter, unidad_hist, id_metodo)
    real(dp), intent(in) :: x0_ini, x1_ini, n_param, tol
    integer, intent(in) :: unidad_hist, id_metodo
    real(dp), intent(out) :: raiz
    integer, intent(out) :: n_iter
    real(dp) :: x_prev, x_curr, f_prev, f_curr, denom, x_next
    integer :: k

    x_prev = x0_ini
    x_curr = x1_ini
    f_prev = F_wien(x_prev, n_param)
    f_curr = F_wien(x_curr, n_param)
    n_iter = 0
    raiz = x_curr

    do k = 1, max_iter_ceros
       denom = f_curr - f_prev
       if (abs(denom) <= tiny(1.0_dp)) then
          stop 'secante: denominador nulo'
       end if
       x_next = x_curr - f_curr * (x_curr - x_prev) / denom
       n_iter = k
       raiz = x_next
       if (unidad_hist /= 0) then
          write(unidad_hist, '(2i6, es24.16)') id_metodo, k, x_next
       end if
       if (abs(x_next - x_curr) < tol) return
       x_prev = x_curr
       f_prev = f_curr
       x_curr = x_next
       f_curr = F_wien(x_curr, n_param)
    end do
  end subroutine secante

end module modulo_ceros
```

##### PROGRAMA

```python
%%writefile programa_wien.f90
program programa_wien
  use modulo_ceros
  implicit none

  real(dp), parameter :: tol = 1.0e-10_dp
  real(dp), parameter :: h_planck = 6.62607015e-34_dp
  real(dp), parameter :: c_luz    = 299792458.0_dp
  real(dp), parameter :: k_boltz  = 1.380649e-23_dp

  real(dp) :: r_bis5, r_new5, r_sec5, r_new5_05
  real(dp) :: r_bis3, r_new3, r_sec3
  real(dp) :: b_wien
  integer  :: it_bis5, it_new5, it_sec5, it_new5_05
  integer  :: it_bis3, it_new3, it_sec3
  integer  :: u_res, u_it, u_b

  ! Abrimos el archivo de historiales para escribir las iteraciones de n = 5
  open(newunit=u_it, file='iteraciones_wien.dat', status='replace', action='write')

  ! Llamamos a los métodos pasando el descriptor de archivo abierto para registrar los pasos
  call biseccion(1.0_dp, 10.0_dp, 5.0_dp, tol, r_bis5, it_bis5, u_it, 1)
  call newton_raphson(10.0_dp, 5.0_dp, tol, r_new5, it_new5, u_it, 2)
  call secante(10.0_dp, 9.0_dp, 5.0_dp, tol, r_sec5, it_sec5, u_it, 3)

  close(u_it)

  ! Caso n=5 con x0 = 0.5 (método 4, sin escribir al archivo de historial general)
  call newton_raphson(0.5_dp, 5.0_dp, tol, r_new5_05, it_new5_05, 0, 4)

  ! Casos para n=3 (no se guardan en el archivo de historial)
  call biseccion(1.0_dp, 10.0_dp, 3.0_dp, tol, r_bis3, it_bis3, 0, 1)
  call newton_raphson(10.0_dp, 3.0_dp, tol, r_new3, it_new3, 0, 2)
  call secante(10.0_dp, 9.0_dp, 3.0_dp, tol, r_sec3, it_sec3, 0, 3)

  ! Escribir resultados resumidos
  open(newunit=u_res, file='resultado_wien.dat', status='replace', action='write')
  write(u_res, '(2i6, es24.16, i6)') 5, 1, r_bis5, it_bis5
  write(u_res, '(2i6, es24.16, i6)') 5, 2, r_new5, it_new5
  write(u_res, '(2i6, es24.16, i6)') 5, 3, r_sec5, it_sec5
  write(u_res, '(2i6, es24.16, i6)') 3, 1, r_bis3, it_bis3
  write(u_res, '(2i6, es24.16, i6)') 3, 2, r_new3, it_new3
  write(u_res, '(2i6, es24.16, i6)') 3, 3, r_sec3, it_sec3
  write(u_res, '(2i6, es24.16, i6)') 5, 4, r_new5_05, it_new5_05
  close(u_res)

  ! Escribir la constante de Wien
  b_wien = (h_planck * c_luz) / (k_boltz * r_new5)
  open(newunit=u_b, file='constante_wien.dat', status='replace', action='write')
  write(u_b, '(3es24.16)') r_new5, r_new3, b_wien
  close(u_b)

end program programa_wien
```

```python
%%writefile modulo_romberg.f90
! ============================================================
! Módulo:     modulo_romberg
! Propósito:  Integración de Romberg con los nuevos
!             integrandos para Stefan-Boltzmann.

module modulo_romberg
  implicit none
  integer, parameter :: dp = kind(1.0d0)

contains

  ! ----------------------------------------------------------
  ! evaluador_f: selecciona el integrando según la opción
  ! op = 1: integral en t in [0, 1] con cambio de variable x = t/(1-t)
  ! op = 2: integral directa en x in [0, 100] sin cambio de variable

  real(dp) function evaluador_f(u, op)
    real(dp), intent(in) :: u
    integer, intent(in) :: op
    real(dp) :: x

    select case (op)
    case (1)  ! u es t en [0, 1]
       if (u <= 0.0_dp .or. u >= 1.0_dp) then
          evaluador_f = 0.0_dp
       else
          x = u / (1.0_dp - u)
          if (x > 100.0_dp) then
             evaluador_f = 0.0_dp
          else if (x < 1.0e-5_dp) then
             evaluador_f = (x**2) * (1.0_dp - 0.5_dp*x + (x**2)/12.0_dp) / ((1.0_dp - u)**2)
          else
             evaluador_f = ((x**3) / (exp(x) - 1.0_dp)) / ((1.0_dp - u)**2)
          end if
       end if

    case (2)  ! u es x en [0, 100]
       x = u
       if (x <= 0.0_dp .or. x > 100.0_dp) then
          evaluador_f = 0.0_dp
       else if (x < 1.0e-5_dp) then
          evaluador_f = (x**2) * (1.0_dp - 0.5_dp*x + (x**2)/12.0_dp)
       else
          evaluador_f = (x**3) / (exp(x) - 1.0_dp)
       end if

    case default
       evaluador_f = 0.0_dp
    end select
  end function evaluador_f

  ! ----------------------------------------------------------
  ! romberg: Calcula la integral usando la tabla S(0:kmax, 0:kmax)
  !          con parada en la diagonal |S(k,k) - S(k-1,k-1)| < tol

  subroutine romberg(a, b, tol, kmax, op, S, k_final, n_eval)
    real(dp), intent(in) :: a, b, tol
    integer, intent(in) :: kmax, op
    real(dp), allocatable, intent(out) :: S(:,:)
    integer, intent(out) :: k_final, n_eval

    integer :: k, j, i
    real(dp) :: h, suma

    allocate(S(0:kmax, 0:kmax))
    S = 0.0_dp

    ! Nivel k = 0
    h = b - a
    S(0, 0) = 0.5_dp * h * (evaluador_f(a, op) + evaluador_f(b, op))
    n_eval = 2
    k_final = kmax

    do k = 1, kmax
       h = h / 2.0_dp
       suma = 0.0_dp
       do i = 1, 2**(k-1)
          suma = suma + evaluador_f(a + real(2*i - 1, dp) * h, op)
       end do

       n_eval = n_eval + 2**(k-1)
       S(k, 0) = 0.5_dp * S(k-1, 0) + h * suma

       do j = 1, k
          S(k, j) = (4.0_dp**j * S(k, j-1) - S(k-1, j-1)) / (4.0_dp**j - 1.0_dp)
       end do

       ! Criterio de parada de la diagonal
       if (abs(S(k, k) - S(k-1, k-1)) < tol) then
          k_final = k
          exit
       end if
    end do

  end subroutine romberg

end module modulo_romberg
```

```python
%%writefile programa_sigma.f90
! ============================================================
! Programa:   programa_sigma
! Propósito:  Calcula I con cambio de variable en [0,1] y sin
!             cambio de variable en [0,100], obteniendo sigma.

program programa_sigma
  use modulo_romberg
  implicit none

  real(dp), parameter :: tol = 1.0e-8_dp
  integer,  parameter :: kmax = 15
  real(dp), parameter :: h_planck = 6.62607015e-34_dp
  real(dp), parameter :: c_luz    = 299792458.0_dp
  real(dp), parameter :: k_boltz  = 1.380649e-23_dp
  real(dp), parameter :: pi_val   = 4.0_dp * atan(1.0_dp)

  real(dp), allocatable :: S(:,:)
  real(dp) :: I_cv, I_sin, sigma_cv, sigma_sin, factor_sb
  integer  :: k_cv, k_sin, ev_cv, ev_sin, u

  ! Prefactor conforme a la celda de verificación del taller
  factor_sb = 2.0_dp * pi_val * (k_boltz**4) / ((h_planck**3) * (c_luz**2))

  ! 1. Con cambio de variable en t in [0, 1] (op = 1)
  call romberg(0.0_dp, 1.0_dp, tol, kmax, 1, S, k_cv, ev_cv)
  I_cv     = S(k_cv, k_cv)
  sigma_cv = factor_sb * I_cv
  deallocate(S)

  ! 2. Sin cambio de variable en x in [0, 100] (op = 2)
  call romberg(0.0_dp, 100.0_dp, tol, kmax, 2, S, k_sin, ev_sin)
  I_sin     = S(k_sin, k_sin)
  sigma_sin = factor_sb * I_sin
  deallocate(S)

  ! Escribir resultado_sigma.dat
  open(newunit=u, file='resultado_sigma.dat', status='replace', action='write')
  write(u, '(2i6, 2es24.16)') k_cv, ev_cv, I_cv, sigma_cv
  write(u, '(2i6, 2es24.16)') k_sin, ev_sin, I_sin, sigma_sin
  close(u)

end program programa_sigma
```

```python
%%writefile maximo_espectro.py
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import CubicSpline
from IPython.display import Image, display

# Constantes exactas del SI
H_PLANCK = 6.62607015e-34
C_LUZ = 299792458.0
K_BOLTZ = 1.380649e-23
T_SOL = 5772.0

AZUL, DORADO, ROJO, GRIS = "#162A54", "#C9972C", "#B02A20", "#7F7F7F"
plt.rcParams.update({"figure.dpi": 150, "axes.grid": True, "grid.alpha": 0.3})


def planck_lambda(lam_m, T):
    """Radiancia espectral B_lambda(lambda, T) [W m^-2 sr^-1 m^-1]."""
    x = (H_PLANCK * C_LUZ) / (lam_m * K_BOLTZ * T)
    val = np.zeros_like(lam_m, dtype=np.float64)
    mask = x < 700.0
    xm = x[mask]
    lm = lam_m[mask]
    val[mask] = (2.0 * H_PLANCK * C_LUZ**2 / (lm**5)) / np.expm1(xm)
    return val


def derivada_central(y, h):
    """Derivada central en los puntos interiores de una malla uniforme de paso h."""
    return (y[2:] - y[:-2]) / (2.0 * h)


def seleccionar_maximo(raices, d2_func, ref_a):
    """Conserva ceros con segunda derivada negativa; si hay varios, elige el más cercano a ref_a."""
    if len(raices) == 0:
        return np.nan
    neg = raices[d2_func(raices) < 0.0]
    candidatos = neg if len(neg) > 0 else raices
    idx = np.argmin(np.abs(candidatos - ref_a))
    return candidatos[idx]


def leer_constante_wien(ruta="constante_wien.dat"):
    with open(ruta, "r", encoding="utf-8") as f:
        for linea in f:
            partes = linea.strip().split()
            if len(partes) >= 3:
                return float(partes[2])
    raise ValueError("No se pudo leer b de constante_wien.dat")


def leer_x3(ruta="resultado_wien.dat"):
    """Lee x3 (n=3, método 2 Newton o método 1 bisección) de resultado_wien.dat."""
    with open(ruta, "r", encoding="utf-8") as f:
        for linea in f:
            partes = linea.strip().split()
            if len(partes) >= 3:
                n_val = int(float(partes[0]))
                met_val = int(float(partes[1]))
                if n_val == 3 and met_val == 2:
                    return float(partes[2])
    raise ValueError("No se encontró x3 en resultado_wien.dat")


def main():
    b_wien = leer_constante_wien("constante_wien.dat")
    x3 = leer_x3("resultado_wien.dat")

    lam_max_nm = (b_wien / T_SOL) * 1e9
    lam_nu_nm = ((H_PLANCK * C_LUZ) / (K_BOLTZ * T_SOL * x3)) * 1e9

    deltas = [100.0, 50.0, 25.0, 12.5]
    filas_salida = []
    datos_25 = None

    for delta in deltas:
        lam_nm = np.arange(100.0, 3000.0 + 0.5 * delta, delta, dtype=np.float64)
        lam_m = lam_nm * 1e-9
        B = planck_lambda(lam_m, T_SOL)
        h_m = delta * 1e-9

        # (a) Mayor valor en la malla
        max_a_nm = lam_nm[np.argmax(B)]

        # (b) Cero del spline de la derivada central
        dB_c = derivada_central(B, h_m)
        lam_int_nm = lam_nm[1:-1]
        s_d = CubicSpline(lam_int_nm, dB_c)
        raices_b = s_d.roots(extrapolate=False)
        max_b_nm = seleccionar_maximo(raices_b, lambda r: s_d(r, 1), max_a_nm)

        # (c) Cero de la derivada del spline de B
        s_B = CubicSpline(lam_nm, B)
        ds_B = s_B.derivative()
        raices_c = ds_B.roots(extrapolate=False)
        max_c_nm = seleccionar_maximo(raices_c, lambda r: ds_B(r, 1), max_a_nm)

        err_a = abs(max_a_nm - lam_max_nm)
        err_b = abs(max_b_nm - lam_max_nm)
        err_c = abs(max_c_nm - lam_max_nm)

        filas_salida.append((delta, err_a, err_b, err_c))

        if np.isclose(delta, 25.0):
            datos_25 = (lam_nm, B)

    # Escribir maximo_espectro.dat
    with open("maximo_espectro.dat", "w", encoding="utf-8") as f:
        for delta, ea, eb, ec in filas_salida:
            f.write(f"{delta:24.16e}{ea:24.16e}{eb:24.16e}{ec:24.16e}\n")

    # Gráfica en espectro_solar.png (malla de 25 nm)
    lam25_nm, B25 = datos_25
    lam_fino_nm = np.linspace(100.0, 3000.0, 1200)
    B_fino = planck_lambda(lam_fino_nm * 1e-9, T_SOL)

    fig, ax = plt.subplots(figsize=(8, 4.5))
    ax.plot(lam_fino_nm, B_fino, color=GRIS, lw=1.5, ls="--", label="Espectro continuo")
    ax.plot(lam25_nm, B25, "o-", ms=3.5, color=AZUL, label=r"Malla $\Delta = 25\ \mathrm{nm}$")
    ax.axvline(lam_max_nm, color=ROJO, lw=1.8, label=rf"$\lambda_{{\max}} = {lam_max_nm:.1f}\ \mathrm{{nm}}$ (ley de Wien)")
    ax.axvline(lam_nu_nm, color=DORADO, lw=1.8, ls="-.", label=rf"$\lambda_{{\nu,\max}} = {lam_nu_nm:.1f}\ \mathrm{{nm}}$ (máx. en frecuencia)")

    ax.set_xlabel(r"Longitud de onda $\lambda$ [nm]")
    ax.set_ylabel(r"$B_\lambda(\lambda, 5772\ \mathrm{K})$ [$\mathrm{W\ m^{-2}\ sr^{-1}\ m^{-1}}$]")
    ax.set_title(r"Espectro solar (cuerpo negro a $T = 5772\ \mathrm{K}$)")
    ax.set_xlim(100.0, 3000.0)
    ax.legend(loc="upper right")
    plt.tight_layout()
    plt.savefig("espectro_solar.png")
    plt.close(fig)


if __name__ == "__main__":
    main()

datos = np.loadtxt("maximo_espectro.dat")
print("Delta [nm] | Error (a) [nm] | Error (b) [nm] | Error (c) [nm]")
for fila in datos:
    print(f"{fila[0]:10.1f} | {fila[1]:14.6e} | {fila[2]:14.6e} | {fila[3]:14.6e}")
    display(Image(filename="espectro_solar.png"))
```

#### PREGUNTA E

1. Costos de los metodos:

Para n = 5, la biseccion de [1, 10] requiere mas iteraciones (convergencai lineal a razon de $1/2$), mientras que, con Newton-Raphson desde $x_0 = 10$. Converge en pocas iteraciones $p \approx 2$ y la secante ($x_0 = 10, x_1 = 9$) logra la proporcion aurea $p \approx 1.618$ usando una sola evaluacion de nueva de F por paso.

Con los errores consecutivos en $e_k=  \vert{}x_k - \alpha\vert{}$ registrados en iteraciones_wien.dat, el orden se estima con $$p \approx \frac{\ln(e_{k+1}/e_k)}{\ln(e_k/e_{k-1})}$$

2. Valor inicial:  ($x_0 = 0.5$ para $n = 5$):

Al partir de $x_0 = 0.5$ con $n = 5$, Newton-Raphson converge a la raíz trivial $x = 0$ (en lugar de $x_5 \approx 4.96511$).

Es raiz exacta de $F(x) = x - 5(1 - e^{-x})$, pues $0 - 5(1 - 1) = 0$. No obstante, no es la respuesta fisica porque en $\lambda_{\max} = \frac{hc}{k_B T x}$, $x = 0$ corresponde a una longitud de onda divergente donde $\lambda_{\max} \to \infty$ y no al pico finito de radciacion termica.


3.Parada prematura en [0, 100] sin cambio de variable:

En [0, 1oo], $\int_0^{100} \frac{x^3}{e^x - 1}\, dx ≈\frac{\pi^4}{15} \approx 6.493939402266828$, pero en los primeros niveles de Romberg los puntos de evaluacion estan lejos del pico, estando alrededor de $x ≈2.82$ o, donde la integracion es despreciable, haciendo que $S_{k,k} - S_{k-1,k-1}$ parezca pequeño o coincida accidentalmete abtes de resolver correctamente la region activa.

Comprobacion adicional: verificar el residuo relativo, exigir que el criterio diagonal se cumpla en dos niveles consecutivos y comprobar la estabilidad al refinar un nivel adicional o dividir el intervalo alrededor del pico.

4. El eslabon mas debil en la parte D:

Al redusir $Δ$ a la mitad los errores de ceri dek spline de la derivada central y cero de la derivada del spline $B_λ$ disminuyen aproximadamente a 4, esto debido a que la derivada central es de orden $\mathcal{O}(\Delta^2)$ y actua como el eslabon mas debil en el proceso

El error de argmax sobre la malla, no baja regular en estas mallas porque esta  acotado por la discretización en pasos de $\Delta$ (del orden de $\Delta$ o $\Delta/2$, dependiendo de qué tan cerca caiga un nodo de $\lambda_{\max}$).

5.Dos máximos (longitud de onda vs. frecuencia):

 El máximo de $B_\lambda$ ocurre en $x_5 \approx 4.965$ con $\lambda_{\max} \approx 502\text{ nm}$ para $T = 5772\text{ K}$, mientras que el máximo de $B_\nu$ ocurre en $x_3 \approx 2.821$ $\lambda_{\nu,\max} = \frac{hc}{k_B T x_3} \approx 883\text{ nm}$.

 No coinciden porque $B_\lambda\,d\lambda = -B_\nu\,d\nu$ y el jacobiano $\left\vert{}\frac{d\nu}{d\lambda}\right\vert{} = \frac{c}{\lambda^2}$ deforma la distribución al cambiar de variable espectral.

 ¿Cuál es "el" color del Sol?

 El color del sol responde a la distribución espectral por longitud de onda/frecuencia ponderada por las curvas de sensibilidad de los conos en el rango visible del ser humano, que es entre $380\text{–}750\text{ nm}$, donde el Sol emite con fuerza, es blanco en el espacio, o ligeramente amarillento desde la tierra por dispersión atmosférica. Ninguno de los dos picos por sí solo resume el color percibido.

```python
# ==================== VERIFICACIÓN DEL TALLER 8 Y 9 (no modificar) ====================
import os
import numpy as np
from scipy.interpolate import CubicSpline as _CS

H_, C_, KB_ = 6.62607015e-34, 299792458.0, 1.380649e-23

def raiz_referencia(n):
    x = float(n)
    for _ in range(200):
        x = n*(1 - np.exp(-x))
    return x

X5, X3 = raiz_referencia(5), raiz_referencia(3)
B_REF = H_*C_/(KB_*X5)
I_REF = np.pi**4/15
S_REF = 2*np.pi*KB_**4/(H_**3*C_**2)*I_REF
ITER_ESPERADAS = {(5,1):37, (5,2):4, (5,3):5,
                  (3,1):37, (3,2):5, (3,3):6}

_avisados = set()
def tabla(dat, fuentes, forma=None):
    for ruta in [dat, *fuentes]:
        if not os.path.isfile(ruta):
            raise ValueError(f"no existe {ruta}; ejecute desde la carpeta del taller")
    antiguos = [f for f in fuentes if os.path.getmtime(dat) < os.path.getmtime(f)]
    if antiguos and dat not in _avisados:
        print(f"[AVISO] {dat} parece anterior a su fuente; vuelva a generar si editó el código.")
        _avisados.add(dat)
    t = np.loadtxt(dat, ndmin=2)
    if forma is not None and t.shape != forma:
        raise ValueError(f"{dat}: forma {t.shape}; se espera {forma}")
    if not np.all(np.isfinite(t)):
        raise ValueError(f"{dat}: contiene NaN o infinitos")
    return t

def enteros(a, nombre):
    if not np.all(a == np.floor(a)):
        raise ValueError(f"{nombre}: se requieren valores enteros")

def resultados_wien():
    t = tabla('resultado_wien.dat', ['modulo_ceros.f90', 'programa_wien.f90'], (7,4))
    enteros(t[:,[0,1,3]], 'n, método e iteraciones')
    claves = [tuple(map(int, fila[:2])) for fila in t[:6]]
    if len(set(claves)) != 6 or set(claves) != set(ITER_ESPERADAS):
        raise ValueError('debe aparecer una vez cada combinación de n=5,3 y método=1,2,3')
    if tuple(t[6,:2]) != (5,4):
        raise ValueError('la fila 7 debe identificar n=5, método=4')
    if np.any((t[:,3] < 1) | (t[:,3] > 100)):
        raise ValueError('conteos fuera de 1 a 100')
    return t

def constantes():
    t = tabla('constante_wien.dat', ['modulo_ceros.f90', 'programa_wien.f90'], (1,3))[0]
    if abs(t[0]-X5)>2e-10 or abs(t[1]-X3)>2e-10:
        raise ValueError('x5 o x3 incorrectos en constante_wien.dat')
    if abs(t[2]-B_REF)/B_REF > 1e-9:
        raise ValueError('constante de Wien incorrecta')
    return t

resultados = []
def caso(nombre, funcion):
    try:
        detalle = funcion()
        ok = True
    except Exception as e:
        ok, detalle = False, str(e)
    resultados.append(ok)
    print(f"[{'PASS' if ok else 'FAIL'}] {nombre}" + (f" -> {detalle}" if detalle else ''))

def c1():
    t = resultados_wien()
    for n, m, raiz, it in t[:6]:
        ref = X5 if n == 5 else X3
        if abs(raiz-ref)>2e-10:
            raise ValueError(f'n={n:g}, método={m:g}: raíz incorrecta')
        esperado = ITER_ESPERADAS[int(n),int(m)]
        if it != esperado:
            raise ValueError(f'n={n:g}, método={m:g}: {it:g} iteraciones; se esperan {esperado}')
    return f'x5={X5:.10f}, x3={X3:.10f}; conteos correctos'

def historia_referencia(m):
    # Recurrencias independientes para verificar los registros de n=5.
    f = lambda x: x-5*(1-np.exp(-x))
    df = lambda x: 1-5*np.exp(-x)
    a,b,prev = 1.0,10.0,10.0
    x = {1:0., 2:10., 3:9.}[m]
    valores=[]
    for _ in range(ITER_ESPERADAS[5,m]):
        if m==1:
            z=(a+b)/2
            if f(a)*f(z)<=0: b=z
            else: a=z
        elif m==2: z=x-f(x)/df(x)
        else: z=x-f(x)*(x-prev)/(f(x)-f(prev))
        valores.append(z)
        prev,x=x,z
    return np.array(valores)

def c2():
    t = tabla('iteraciones_wien.dat', ['modulo_ceros.f90','programa_wien.f90'])
    if t.shape[1]!=3: raise ValueError('iteraciones_wien.dat debe tener tres columnas')
    enteros(t[:,:2], 'método y contador')
    if set(t[:,0])!={1,2,3}: raise ValueError('faltan métodos o hay identificadores inválidos')
    for m in range(1,4):
        filas=t[t[:,0]==m]
        ref=historia_referencia(m)
        if len(filas)!=len(ref): raise ValueError(f'método {m}: número de registros incorrecto')
        if not np.array_equal(filas[:,1], np.arange(1,len(ref)+1)):
            raise ValueError(f'método {m}: use contadores consecutivos desde 1, en orden')
        if not np.allclose(filas[:,2],ref,atol=2e-13,rtol=2e-14):
            raise ValueError(f'método {m}: trayectoria incorrecta o pocas cifras en el archivo')
    return 'tres historiales completos'

def c3():
    x5,x3,b=constantes()
    t=resultados_wien()
    if abs(t[6,2])>1e-8: raise ValueError('Newton desde 0.5 debe llegar a la raíz trivial')
    return f'b={b:.9e} m K; raíz trivial comprobada'

def c4():
    t=tabla('resultado_sigma.dat', ['modulo_romberg.f90','programa_sigma.f90'], (2,4))
    enteros(t[:,:2], 'nivel y evaluaciones de Romberg')
    for k,ne,I,s in t:
        if k<0 or k>50 or ne<2:
            raise ValueError('nivel o conteo de Romberg inválido')
        if int(ne)!=2**int(k)+1:
            raise ValueError('use nivel k=0 para el trapecio inicial y cuente nodos únicos: 2**k+1')
        factor=2*np.pi*KB_**4/(H_**3*C_**2)
        if not np.isclose(s,factor*I,rtol=1e-8,atol=1e-30):
            raise ValueError('sigma no corresponde a la integral de su fila')
    if abs(t[0,2]-I_REF)>1e-8 or abs(t[0,3]-S_REF)/S_REF>1e-8:
        raise ValueError('la integral transformada o sigma no alcanzan la precisión solicitada')
    prematura=100/6*(4*50**3/np.expm1(50)+100**3/np.expm1(100))
    if t[1,0]!=1 or not np.isclose(t[1,2],prematura,rtol=1e-6,atol=1e-18):
        raise ValueError('fila 2: debe mostrar la parada prematura en k = 1')
    return f'sigma={t[0,3]:.9e} W/m²/K⁴; parada prematura reproducida'

def c5():
    constantes()
    t=tabla('maximo_espectro.dat', ['maximo_espectro.py','constante_wien.dat'], (4,4))
    if not np.array_equal(t[:,0],[100.,50.,25.,12.5]):
        raise ValueError('use, en este orden, pasos de 100, 50, 25 y 12.5 nm')
    if np.any(t[:,1:]<0): raise ValueError('los errores absolutos no pueden ser negativos')
    ref=B_REF/5772*1e9
    for fila in t:
        dl=fila[0]
        l=np.arange(100.,3000.+dl/2,dl)
        B=1/l**5/np.expm1(H_*C_/(l*1e-9*KB_*5772))
        la=l[np.argmax(B)]
        def cero(s):
            r=s.roots(extrapolate=False)
            r=r[np.isfinite(r)]
            r=r[s(r,1)<0]
            if len(r)==0: raise ValueError('no se encontró candidato a máximo')
            return r[np.argmin(abs(r-la))]
        esperados=[abs(la-ref),
            abs(cero(_CS(l[1:-1],(B[2:]-B[:-2])/(2*dl)))-ref),
            abs(cero(_CS(l,B).derivative())-ref)]
        for nombre,val,esp in zip(['a','b','c'],fila[1:],esperados):
            if abs(val-esp)>1e-4+0.01*esp:
                raise ValueError(f'paso {dl:g} nm, método {nombre}: error incompatible')
    if not os.path.isfile('espectro_solar.png'):
        raise ValueError('falta espectro_solar.png')
    return 'cuatro mallas correctas'

caso('1. Raíces e iteraciones de los tres métodos', c1)
caso('2. Historiales de iteración y convergencia', c2)
caso('3. Constante de Wien y Newton desde x0=0.5', c3)
caso('4. Stefan–Boltzmann y parada prematura', c4)
caso('5. Máximo del espectro solar en cuatro mallas', c5)
print(f'\nResultado: {sum(resultados)} de {len(resultados)} PASS')
```
